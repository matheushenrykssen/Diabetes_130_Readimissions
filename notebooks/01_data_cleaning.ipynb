## Data Cleaning

```python
df_2 = pd.read_csv(path_file_2)
df_2
df_2['admission_type_id'].unique()
# IDS_mapping file, have 3 tables concatenated
```

### IDS_mapping data

```python
# Discover the position of the first occurrence of "discharge_disposition_id" in the "admission_type_id" column
# in 9
id_position_1 = (
    df_2[df_2["admission_type_id"] == "discharge_disposition_id"].index[0]
)
id_position_1
```

```python
# Repeat the step for another occurrence
# in 41
id_position_2 = (
    df_2[df_2["admission_type_id"] == "admission_source_id"].index[0]
)
id_position_2
```

```python
# Separate the three tables in the IDS_mapping file based on the discovered positions
admission_type = df_2.iloc[0:id_position_1].copy()
admission_type 
```

```python
admission_type = admission_type.dropna()
admission_type
```

```python
# Rename columns 
admission_type.columns = ["admission_type_id", "admission_type_description"]
admission_type
```

```python
# Return the data type of the 'admission_type_id' to int
admission_type['admission_type_id'] = admission_type['admission_type_id'].astype(int)
admission_type.dtypes
```

```python
# For discharge_disposition
discharge = df_2.iloc[id_position_1 + 1:id_position_2].copy()
discharge = discharge.dropna()
discharge.columns = ['discharge_disposition_id', 'discharge_disposition_description']
discharge['discharge_disposition_id'] = discharge['discharge_disposition_id'].astype(int)
discharge.dtypes
```

```python
# For admission_source
source = df_2.iloc[id_position_2 + 1:].copy()
source = source.dropna()
source.columns = ['admission_source_id', 'admission_source_description']
source['admission_source_id'] = source['admission_source_id'].astype(int)
source.dtypes
```
