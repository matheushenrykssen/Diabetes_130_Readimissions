# Medication

```python
# Create a dimension for medications
medication_columns = [
    'metformin',
    'repaglinide',
    'nateglinide',
    'chlorpropamide',
    'glimepiride',
    'acetohexamide',
    'glipizide',
    'glyburide',
    'tolbutamide',
    'pioglitazone',
    'rosiglitazone',
    'acarbose',
    'miglitol',
    'troglitazone',
    'tolazamide',
    'examide',
    'citoglipton',
    'insulin',
    'glyburide-metformin',
    'glipizide-metformin',
    'glimepiride-pioglitazone',
    'metformin-rosiglitazone',
    'metformin-pioglitazone'
]

dim_medication = (
    diabetes[medication_columns]
    .drop_duplicates()
    .reset_index(drop=True)
)
```

```python
dim_medication.insert(
    0,
    'medication_id',
    range(1, len(dim_medication)+ 1)
)
```

```python
# dim_medication
medication_lookup = (
    dim_medication
    .set_index(medication_columns)['medication_id']
    .to_dict()
)
```

```python
# Rename columns for SQL Server
dim_medication= dim_medication.rename(columns={
    'metformin': 'metformin_status',
    'repaglinide': 'repaglinide_status',
    'nateglinide': 'nateglinide_status',
    'chlorpropamide': 'chlorpropamide_status',
    'glimepiride': 'glimepiride_status',
    'acetohexamide': 'acetohexamide_status',
    'glipizide': 'glipizide_status',
    'glyburide': 'glyburide_status',
    'tolbutamide': 'tolbutamide_status',
    'pioglitazone': 'pioglitazone_status',
    'rosiglitazone': 'rosiglitazone_status',
    'acarbose': 'acarbose_status',
    'miglitol': 'miglitol_status',
    'troglitazone': 'troglitazone_status',
    'tolazamide': 'tolazamide_status',
    'examide': 'examide_status',
    'citoglipton': 'citoglipton_status',
    'insulin': 'insulin_status',
    'glyburide-metformin': 'glyburide_metformin_status',
    'glipizide-metformin': 'glipizide_metformin_status',
    'glimepiride-pioglitazone': 'glimepiride_pioglitazone_status',
    'metformin-rosiglitazone': 'metformin_rosiglitazone_status',
    'metformin-pioglitazone': 'metformin_pioglitazone_status'
})
```

```python
dim_medication
```

```python
medication_status_columns = [
    'metformin_status',
    'repaglinide_status',
    'nateglinide_status',
    'chlorpropamide_status',
    'glimepiride_status',
    'acetohexamide_status',
    'glipizide_status',
    'glyburide_status',
    'tolbutamide_status',
    'pioglitazone_status',
    'rosiglitazone_status',
    'acarbose_status',
    'miglitol_status',
    'troglitazone_status',
    'tolazamide_status',
    'examide_status',
    'citoglipton_status',
    'insulin_status',
    'glyburide_metformin_status',
    'glipizide_metformin_status',
    'glimepiride_pioglitazone_status',
    'metformin_rosiglitazone_status',
    'metformin_pioglitazone_status'
]
```

```python
dim_medication.columns
```

```python
# Temp dataframe
medication_merge = diabetes.merge(
    dim_medication,
    left_on=medication_columns,
    right_on=medication_status_columns,
    how='left'
)
```

# Readmission

```python
dim_readmission = (
    diabetes[['readmitted']]
    .drop_duplicates()
    .sort_values('readmitted')
    .reset_index(drop=True)
)

dim_readmission.insert(
    0,
    'readmission_id',
    range(1, len(dim_readmission) + 1)
)

dim_readmission
```

```python
readmission_lookup = (
    dim_readmission
    .set_index('readmitted')['readmission_id']
    .to_dict()
)
```

# Admission Source

```python
dim_admission_source = adm_source.copy()
dim_admission_source
```

# Discharge Disposition

# Fact

```python
fact_hospitalization = diabetes[
    [
        'encounter_id',
        'patient_nbr',
        'admission_type_id',
        'admission_source_id',
        'discharge_disposition_id',
        'age',
        'gender',
        'diabetesMed',
        'change',
        'A1Cresult',
        'max_glu_serum',
        'time_in_hospital',
        'num_medications',
        'number_emergency',
        'number_inpatient',
        'number_outpatient'
    ]
].copy()
```

```python
fact_hospitalization = fact_hospitalization.rename(columns={
    'diabetesMed' : 'diabetes_med',
    'change' : 'medication_changed',
    'A1Cresult' : 'a1c_result'
})
```

```python
fact_hospitalization['diagnosis_id'] = (
    diabetes['diag_1'].map(diagnosis_lookup)
)
```

```python
fact_hospitalization['readmission_id'] = (
    diabetes['readmitted']
    .map(readmission_lookup)
)
```

```python
fact_hospitalization['medication_id'] = medication_merge['medication_id']
fact_hospitalization
```

```python
# organize columns from SQL Server

fact_hospitalization = fact_hospitalization[
    [
        'encounter_id',
        'patient_nbr',
        'admission_type_id',
        'admission_source_id',
        'discharge_disposition_id',
        'diagnosis_id',
        'medication_id',
        'readmission_id',
        'age',
        'gender',
        'diabetes_med',
        'medication_changed',
        'a1c_result',
        'max_glu_serum',
        'time_in_hospital',
        'num_medications',
        'number_emergency',
        'number_inpatient',
        'number_outpatient'
    ]
]
```

```python
# convert to BIT values

 fact_hospitalization["diabetes_med"] = (
    fact_hospitalization["diabetes_med"]
    .map({
        "Yes": 1,
        "No": 0
    })
    .astype("int8")
) 

fact_hospitalization["medication_changed"] = (
    fact_hospitalization["medication_changed"]
    .map({
        "Ch": 1,
        "No": 0
    })
    .astype("int8")
)
```

### Left JOIN Diagnosis Test

```python
# Check if all diag_1 have a matching
validation_diag = (
    fact_hospitalization[['diagnosis_id']]
    .merge(
        dim_diagnosis[['diagnosis_id', 'icd9_code']],
        on='diagnosis_id',
        how='left'
    )
)

(diabetes['diag_1'] == validation_diag['icd9_code']).all()
```

### Readmission Test

```python
fact_hospitalization["readmission_id"].isna().sum()
```

```python
validation_readmission = (
    fact_hospitalization[['readmission_id']]
    .merge(
        dim_readmission[['readmission_id', 'readmitted']],
        on='readmission_id',
        how='left'
    )
)
(diabetes['readmitted'] == validation_readmission['readmitted']).all()
```

### Medication Test

```python
fact_hospitalization['medication_id'].isna().sum()
```

```python
validation_meddication = (
    fact_hospitalization[['medication_id']]
    .merge(
        dim_medication,
        on='medication_id',
        how='left'
    )
)

compare = (
    diabetes[medication_columns].reset_index(drop=True)
    ==
    validation_meddication[medication_status_columns]
        .rename(columns=dict(zip(medication_status_columns, medication_columns)))
        .reset_index(drop=True)
)

compare.all().all()
```
