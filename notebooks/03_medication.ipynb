# Library and Loads

```python
import pandas as pd

filepath_1 = r'C:\Users\henry\Desktop\Projects\Diabetes_130_US\Dataset\diabetic_data.csv'
diabetes = pd.read_csv(filepath_1)

filepath_2 = r'C:\Users\henry\Desktop\Projects\Diabetes_130_US\Dataset\Dim_&_Fact\dim_medication.csv'
dim_medication = pd.read_csv(filepath_2)
```

# Medication

```python
# Get medication columns from the dimension table
medication_columns = []

for column in dim_medication.columns:

    if column.endswith('_status'):

        original_name = (
            column
            .replace('_status', '')
            .replace('_', '-')
        )

        medication_columns.append(original_name)

# Build medication usage table
medication_usage = []

for medication in medication_columns:

    patients = (diabetes[medication] != 'No').sum()

    usage_rate = round(
        patients / len(diabetes) * 100,
        2
    )

    medication_usage.append({
        'medication_name': medication,
        'patients': patients,
        'usage_rate': usage_rate
    })

fact_medication_usage = (
    pd.DataFrame(medication_usage)
    .sort_values('patients', ascending=False)
    .reset_index(drop=True)
)

fact_medication_usage['medication_name'] = (
    fact_medication_usage['medication_name']
    .str.replace('-', ' ', regex=False)
    .str.title()
)

output_path = (
    r'C:\Users\henry\Desktop\Projects\Diabetes_130_US\Dataset\Dim_&_Fact'
    r'\fact_medication_usage.csv'
)

fact_medication_usage.to_csv(
    output_path,
    index=False
)

print(fact_medication_usage)
```
